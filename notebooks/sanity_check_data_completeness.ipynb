```python
import pandas as pd
raw = pd.read_csv("../data/eval/ragas_scores_raw.csv")
agentic = raw[raw["pipeline"] == "agentic"]
nan_rows = agentic[agentic[["faithfulness","context_precision","context_recall","answer_correctness"]].isna().any(axis=1)]
print(nan_rows["answer"].apply(lambda x: len(str(x))).describe())
print((nan_rows["answer"].astype(str).str.strip() == "").sum(), "empty answers out of", len(nan_rows))
print(nan_rows[nan_rows["answer"].str.len() < 10]["answer"].tolist())
```
